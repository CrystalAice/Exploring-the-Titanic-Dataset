# Project Overview


---

In this project I explored and analysed the train.csv (Titanic passenger) dataset downloaded from [Kaggle Titanic](https://https://www.kaggle.com/c/titanic/data).

I dicovered that the data has some missing values specifically in the following columns:

*   Age
*   Cabin
*   Embarked

The data is made up of a mixture of strings, integers and floats. However, the  Age column was changed from floats to integers for the analysis to be easier.

The NULL Age column values were filled with (Age = 0) and the rest of the columns with NULL values with 'Unknown'. This allowed for an easy way to find the number survivors per age.
